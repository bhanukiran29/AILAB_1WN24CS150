```python
# Unification Algorithm in First Order Logic


def is_variable(x):
    return isinstance(x, str) and x.startswith("?")


def apply_substitution(term, subst):
    if is_variable(term):
        if term in subst:
            return apply_substitution(subst[term], subst)
        return term

    if isinstance(term, tuple):
        return tuple(apply_substitution(x, subst) for x in term)

    return term


def occurs_check(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, x, subst) for x in term)

    return False


def unify(x, y, subst):

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    # Step 1: If both are identical
    if x == y:
        return subst

    # Step 2: If x is a variable
    if is_variable(x):
        if occurs_check(x, y, subst):
            return None

        subst[x] = y
        return subst

    # Step 3: If y is a variable
    if is_variable(y):
        if occurs_check(y, x, subst):
            return None

        subst[y] = x
        return subst

    # Step 4: Check predicate/function symbols
    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return None

        # Step 5: Check number of arguments
        if len(x) != len(y):
            return None

        # Step 6: Unify corresponding arguments
        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    # Step 7: Otherwise, unification fails
    return None


# Main Program
print("UNIFICATION ALGORITHM")
print("1. Successful Unification")
print("2. Failed Unification")

choice = int(input("Enter your choice: "))

match choice:

    case 1:
        expr1 = ("Eats", "?x", "Apple")
        expr2 = ("Eats", "Bhanu", "?y")

    case 2:
        expr1 = ("Knows", "John", "?x")
        expr2 = ("Knows", "?x", "Elizabeth")

    case _:
        print("Invalid choice!")
        exit()


print("\nExpression 1:", expr1)
print("Expression 2:", expr2)

result = unify(expr1, expr2, {})

if result is not None:
    print("\nUnification Successful!")
    print("Substitution Set:")

    for variable, value in result.items():
        print(variable, "=", value)

    print("\nUnified Expression:")
    print(apply_substitution(expr1, result))

else:
    print("\nUnification Failed!")

print(" ")
print("Executed by MBK")
```
