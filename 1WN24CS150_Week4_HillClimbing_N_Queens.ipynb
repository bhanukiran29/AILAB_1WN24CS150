```python
def calculate_cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    neighbors = []
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            new_state = list(state)
            new_state[i], new_state[j] = new_state[j], new_state[i]
            neighbors.append((tuple(new_state), (i, j)))

    return neighbors


def hill_climbing(initial):
    current = initial

    while True:
        current_cost = calculate_cost(current)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("Goal Reached!")
            return current

        neighbors = get_neighbors(current)

        print("\nNeighbor States:")

        best_state = current
        best_cost = current_cost
        best_swap = None

        for neighbor, swap in neighbors:
            cost = calculate_cost(neighbor)

            print("Swap", swap, ":", neighbor, "Cost:", cost)

            if cost < best_cost:
                best_state = neighbor
                best_cost = cost
                best_swap = swap

        if best_state == current:
            print("No better neighbor found. Algorithm terminated.")
            return current

        print("\nSelected Swap:", best_swap)
        print("Selected State:", best_state)
        print("Selected Cost:", best_cost)

        current = best_state


initial_state = (3, 1, 2, 0)

print("HILL CLIMBING - 4 QUEENS")
print("Initial State:", initial_state)
print("Initial Cost:", calculate_cost(initial_state))

solution = hill_climbing(initial_state)

print("\nFinal Solution:", solution)
print("Final Cost:", calculate_cost(solution))
print("")
print("Executed by Madisetty Bhanu Kiran")
```
