```python
import random
import math


def calculate_cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbor(state):
    new_state = list(state)

    i, j = random.sample(range(len(state)), 2)

    new_state[i], new_state[j] = new_state[j], new_state[i]

    return tuple(new_state)


def simulated_annealing(initial):
    current = initial
    temperature = 100.0
    cooling_rate = 0.95
    min_temperature = 0.01

    while temperature > min_temperature:
        current_cost = calculate_cost(current)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)
        print("Temperature:", round(temperature, 2))

        if current_cost == 0:
            print("Goal Reached!")
            return current

        next_state = get_neighbor(current)
        next_cost = calculate_cost(next_state)

        delta = current_cost - next_cost

        if delta > 0:
            current = next_state
            print("Better neighbor accepted:", current)
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = next_state
                print("Worse or equal neighbor accepted:", current)
            else:
                print("Neighbor rejected.")

        temperature *= cooling_rate

    if calculate_cost(current) == 0:
        print("Goal Reached!")
    else:
        print("Temperature reached minimum.")

    return current


# Main program
initial_state = (3, 1, 2, 0)

print("SIMULATED ANNEALING - 4 QUEENS")
print("Initial State:", initial_state)
print("Initial Cost:", calculate_cost(initial_state))

solution = simulated_annealing(initial_state)

print("\nFinal State:", solution)
print("Final Cost:", calculate_cost(solution))

print("")
print("Executed by Madisetty Bhanu Kiran")
```
