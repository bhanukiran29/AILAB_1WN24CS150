```python

facts = {
    ("American", "Robert"),
    ("Enemy", "CountryA", "America"),
    ("Owns", "CountryA", "T1"),
    ("Missile", "T1")
}

rules = [
    ([("Enemy", "?x", "America")],
     ("Hostile", "?x")),

    ([("Missile", "?x")],
     ("Weapon", "?x")),

    ([("Missile", "?x")],
     ("Sells", "Robert", "?x", "CountryA")),

    ([("American", "?x"),
      ("Weapon", "?y"),
      ("Sells", "?x", "?y", "?z"),
      ("Hostile", "?z")],
     ("Criminal", "?x"))
]


def match(pattern, fact, substitution):
    if pattern[0] != fact[0]:
        return None

    new_sub = substitution.copy()

    if len(pattern) != len(fact):
        return None

    for p, f in zip(pattern[1:], fact[1:]):
        if p.startswith("?"):
            if p in new_sub and new_sub[p] != f:
                return None
            new_sub[p] = f
        elif p != f:
            return None

    return new_sub


def apply_rule(premises, conclusion, facts):
    substitutions = [{}]

    for premise in premises:
        new_substitutions = []

        for sub in substitutions:
            for fact in facts:
                result = match(premise, fact, sub)

                if result is not None:
                    new_substitutions.append(result)

        substitutions = new_substitutions

    derived = set()

    for sub in substitutions:
        new_fact = tuple(
            sub.get(item, item) for item in conclusion
        )

        derived.add(new_fact)

    return derived


def forward_chaining(facts, rules):
    iteration = 1

    while True:
        new_facts = set()

        for premises, conclusion in rules:
            derived = apply_rule(premises, conclusion, facts)
            new_facts.update(derived - facts)

        if not new_facts:
            break

        print("\nIteration", iteration)

        for fact in sorted(new_facts):
            print("Derived:", fact)

        facts.update(new_facts)
        iteration += 1

    print("\nFinal Knowledge Base:")

    for fact in sorted(facts):
        print(fact)

    query = ("Criminal", "Robert")

    if query in facts:
        print("\nSUCCESS: Robert is a Criminal.")
    else:
        print("\nFAILURE: Query cannot be proved.")


print("FOL - FORWARD CHAINING")
print("\nInitial Facts:")

for fact in sorted(facts):
    print(fact)

forward_chaining(facts, rules)


print("")
print("executed by mbk")
```
