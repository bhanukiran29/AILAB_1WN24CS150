```python
from heapq import heappush, heappop

initial = (
    2, 8, 3,
    1, 6, 4,
    0, 7, 5
)

goal = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


# Print the puzzle
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# Generate possible moves
def get_neighbors(state):
    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = []

    if row > 0:
        moves.append(zero - 3)       # UP

    if row < 2:
        moves.append(zero + 3)       # DOWN

    if col > 0:
        moves.append(zero - 1)       # LEFT

    if col < 2:
        moves.append(zero + 1)       # RIGHT

    neighbors = []

    for position in moves:
        new_state = list(state)
        new_state[zero], new_state[position] = \
            new_state[position], new_state[zero]

        neighbors.append(tuple(new_state))

    return neighbors


# Case 1: Number of misplaced tiles
def misplaced_tiles(state, goal):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Case 2: Manhattan Distance
def manhattan_distance(state, goal):
    distance = 0

    for tile in range(1, 9):

        current_position = state.index(tile)
        goal_position = goal.index(tile)

        current_row = current_position // 3
        current_col = current_position % 3

        goal_row = goal_position // 3
        goal_col = goal_position % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# A* Search
def a_star(initial, goal, heuristic):

    open_list = []

    g = 0
    h = heuristic(initial, goal)
    f = g + h

    heappush(open_list, (f, g, initial, [initial]))

    closed = set()

    while open_list:

        f, g, current, path = heappop(open_list)

        if current in closed:
            continue

        closed.add(current)

        if current == goal:
            return path

        for next_state in get_neighbors(current):

            if next_state in closed:
                continue

            new_g = g + 1
            new_h = heuristic(next_state, goal)
            new_f = new_g + new_h

            heappush(
                open_list,
                (new_f, new_g, next_state, path + [next_state])
            )

    return None


# Main Program
print("========== 8-PUZZLE A* SEARCH ==========")

print("\nInitial State:")
print_state(initial)

print("Goal State:")
print_state(goal)

print("Choose Heuristic:")
print("1. Number of Misplaced Tiles")
print("2. Manhattan Distance")

choice = int(input("\nEnter your choice (1 or 2): "))

# Switch-case type using match-case
match choice:

    case 1:
        print("\nHeuristic Selected: Misplaced Tiles")

        solution = a_star(
            initial,
            goal,
            misplaced_tiles
        )

    case 2:
        print("\nHeuristic Selected: Manhattan Distance")

        solution = a_star(
            initial,
            goal,
            manhattan_distance
        )

    case _:
        print("Invalid choice!")
        solution = None


# Display solution
if solution is not None:

    print("\n========== SOLUTION ==========")
    print("Number of moves:", len(solution) - 1)

    for i, state in enumerate(solution):

        print("\nStep", i)
        print_state(state)

else:

    print("\nNo solution found.")
```
