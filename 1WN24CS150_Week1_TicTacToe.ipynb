```python
def print_board(board):
    """Prints the current state of the board in a clean format."""
    for i, row in enumerate(board):
        print(f" {row[0]} | {row[1]} | {row[2]} ")
        if i < 2:
            print("---|---|---")


def check_win(board, player):
    """Checks rows, columns, and diagonals for a winning line."""
    # Check rows and columns
    for i in range(3):
        if all(board[i][j] == player for j in range(3)) or \
           all(board[j][i] == player for j in range(3)):
            return True

    # Check diagonals
    if board[0][0] == player and board[1][1] == player and board[2][2] == player:
        return True
    if board[0][2] == player and board[1][1] == player and board[2][0] == player:
        return True

    return False


def play_game():
    """Main function to handle the game loop and user input."""
    # Initialize game state
    board = [[" " for _ in range(3)] for _ in range(3)]
    current_player = "X"
    moves_count = 0

    print("Welcome to Tic-Tac-Toe!")
    print("Enter your moves using row and column numbers (0, 1, or 2).\n")

    while moves_count < 9:
        print_board(board)
        print(f"\nPlayer {current_player}'s turn.")

        # Get and validate input format
        try:
            row = int(input("Enter row (0-2): "))
            col = int(input("Enter column (0-2): "))
        except ValueError:
            print("Invalid input! Please enter numbers only.\n")
            continue

        # Check if the coordinates are within bounds
        if row not in range(3) or col not in range(3):
            print("Invalid coordinates! Out of bounds (0-2). Try again.\n")
            continue

        # Check if the position is already taken
        if board[row][col] != " ":
            print("That cell is already occupied. Try again.\n")
            continue

        # Make the move
        board[row][col] = current_player
        moves_count += 1

        # Check for a winner
        if check_win(board, current_player):
            print_board(board)
            print(f"\nGame Over! Player {current_player} wins! 🎉")
            return

        # Switch players
        current_player = "O" if current_player == "X" else "X"
        print()  # Clear line spacing

    # If the loop finishes without a winner
    print_board(board)
    print("\nGame Over! It's a tie! 🤝")


# Run the game
if __name__ == "__main__":
    play_game()
```
