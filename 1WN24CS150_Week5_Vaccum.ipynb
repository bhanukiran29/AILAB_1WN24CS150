```python
# Vacuum Cleaner Agent - Week 1 AI Lab

# Two locations: A and B
# Each location can be Dirty or Clean


# -------------------------------------------------
# Simple Reflex Agent
# -------------------------------------------------

print("----- Simple Reflex Agent -----")

environment = {
    'A': 'Dirty',
    'B': 'Dirty'
}

location = 'A'

for step in range(4):

    status = environment[location]

    print(f"Step {step}: Location = {location}, Status = {status}")

    if status == 'Dirty':
        print("Action: Suck")
        environment[location] = 'Clean'

    elif location == 'A':
        print("Action: Move Right")
        location = 'B'

    else:
        print("Action: Move Left")
        location = 'A'

print("Final Environment:", environment)


# -------------------------------------------------
# Goal-Based Agent
# -------------------------------------------------

print("\n----- Goal-Based Agent -----")

environment = {
    'A': 'Dirty',
    'B': 'Dirty'
}

goal = {
    'A': 'Clean',
    'B': 'Clean'
}

location = 'A'
step = 0

while environment != goal:

    status = environment[location]

    print(f"Step {step}: Location = {location}, Status = {status}")

    if status == 'Dirty':
        print("Action: Suck")
        environment[location] = 'Clean'

    elif location == 'A':
        print("Action: Move Right")
        location = 'B'

    else:
        print("Action: Move Left")
        location = 'A'

    step += 1

print("Goal Reached!")
print("Final Environment:", environment)
```
